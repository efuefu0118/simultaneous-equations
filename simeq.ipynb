```python
import sympy as sp
import random
sp.init_printing(use_unicode=False, wrap_line=True)
```

```python
from sympy import symbols, expand, latex
x, y, a, b, p, q = symbols('x, y, a, b, p, q')
```

```python
def latex():
    fhalf, shalf = simeq()
    return (
        r"\noindent" "\n"
        r"\begin{minipage}[t]{0.46\textwidth}" "\n"
        r"\begin{enumerate}[label=\arabic*.,leftmargin=8mm,itemsep=25mm,topsep=20mm]" "\n"
        rf"{"\n".join(fhalf)}" "\n"
        r"\end{enumerate}" "\n"
        r"\end{minipage}%" "\n"
        r"\hfill\vrule\hfill"
        r"\begin{minipage}[t]{0.46\textwidth}" "\n"
        r"\begin{enumerate}[start = 8,label=\arabic*.,leftmargin=8mm,itemsep=25mm,topsep=20mm]" "\n"
        rf"{"\n".join(shalf)}" "\n"
        r"\end{enumerate}" "\n"
        r"\end{minipage}%" "\n"
    )
```

```python
def simeq():
    problems = []
    cand = [n for n in range(-9,10) if n != 0]
    for i in range(14):
        while True:
            a = random.choice(cand)
            b = random.choice(cand)
            c = random.choice(cand)
            d = random.choice(cand)
            x = random.randint(-5,5)
            y = random.randint(-5,5)
            if x * y == 0 or (a*d - b*c) == 0:
                continue
            p = a * x + b * y
            q = c * x + d * y
            problem_latex = f"{a}x + {b}y = {p} \\\\ {c}x + {d}y = {q}"
            problem_latex = problem_latex.replace("+ -", "- ")
            problem_latex = problem_latex.replace("1x", "x")
            problem_latex = problem_latex.replace("1y", "y")
            problems.append(rf"\item $\begin{{cases}} {problem_latex} \end{{cases}}$")
            break
    fhalf = problems[:7]
    shalf = problems[7:]
    return (fhalf, shalf)
```

```python
for i in range(1,10): #名前に関しては0から始まるのは気持ち悪いですね...
    with open(f"simeq{i}.tex", "w") as f:
        f.write(latex())
    print(f"ファイル{i}を作成しました")
```
